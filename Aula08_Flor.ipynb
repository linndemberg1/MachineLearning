### Modelo de árvore de decisão capaz de prever a espécie da flor (Setosa, Versicolor ou Virginica) com base nas características das pétalas e sépalas.

```python
# -------------------------------------------------------------
# Exemplo 2: Árvore de Decisão - Classificação das Flores Íris
# Autor: Prof. Lindenberg Andrade
# -------------------------------------------------------------

# 1️⃣ Importação das bibliotecas necessárias
from sklearn.datasets import load_iris                 # Base de dados Iris
from sklearn.model_selection import train_test_split   # Separar treino e teste
from sklearn.tree import DecisionTreeClassifier, plot_tree  # Modelo e visualização
from sklearn.metrics import accuracy_score, classification_report  # Avaliação
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# 2️⃣ Carregando o dataset Iris
# -------------------------------------------------------------
# O dataset Iris é um clássico do Machine Learning.
# Ele possui 150 amostras de flores de 3 espécies:
#  - Setosa
#  - Versicolor
#  - Virginica
# Cada amostra contém 4 atributos numéricos:
#  - comprimento da sépala
#  - largura da sépala
#  - comprimento da pétala
#  - largura da pétala

iris = load_iris()

# Mostrando os nomes das classes e atributos
print("Espécies:", iris.target_names)
print("Atributos:", iris.feature_names)

# -------------------------------------------------------------
# 3️⃣ Criando variáveis de entrada (X) e saída (y)
# -------------------------------------------------------------
X = iris.data   # características das flores
y = iris.target # rótulos (0 = Setosa, 1 = Versicolor, 2 = Virginica)

# -------------------------------------------------------------
# 4️⃣ Dividindo os dados em treino e teste
# -------------------------------------------------------------
# Treinamento com 70% dos dados e teste com 30%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# -------------------------------------------------------------
# 5️⃣ Criando e treinando o modelo de Árvore de Decisão
# -------------------------------------------------------------
# criterion='entropy' → usa entropia para medir impureza
# max_depth=None → deixa a árvore crescer até separar completamente as classes
modelo = DecisionTreeClassifier(criterion='entropy', random_state=42)
modelo.fit(X_train, y_train)

# -------------------------------------------------------------
# 6️⃣ Fazendo previsões no conjunto de teste
# -------------------------------------------------------------
y_pred = modelo.predict(X_test)

# -------------------------------------------------------------
# 7️⃣ Avaliando o desempenho do modelo
# -------------------------------------------------------------
acc = accuracy_score(y_test, y_pred)
print(f"Acurácia do modelo: {acc:.2f}")
print("\nRelatório de classificação:\n", classification_report(y_test, y_pred, target_names=iris.target_names))

# -------------------------------------------------------------
# 8️⃣ Visualizando a Árvore de Decisão
# -------------------------------------------------------------
plt.figure(figsize=(16,10))
plot_tree(
    modelo,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Árvore de Decisão - Classificação das Flores Íris", fontsize=14)
plt.show()
```
