Árvore de Decisão “Devo ir à praia?”

```python
# ---------------------------------------------
# Árvore de Decisão: "Devo ir à praia?"
# Autor: Prof. Lindenberg Andrade
# ---------------------------------------------
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Dados simulados
# Temperatura (0 = Baixa, 1 = Alta)
# Chuva (0 = Sim, 1 = Não)
dados = pd.DataFrame({
    'Temperatura': [1, 1, 0, 0, 1, 0],
    'Chuva': [0, 1, 1, 0, 1, 0],
    'Ir_Praia': ['Não', 'Sim', 'Não', 'Não', 'Sim', 'Não']
})

# Separar atributos e rótulo
X = dados[['Temperatura', 'Chuva']]
y = dados['Ir_Praia']

# Criar modelo
modelo = DecisionTreeClassifier(criterion='entropy', max_depth=2, random_state=0)
modelo.fit(X, y)

# Plotar árvore
plt.figure(figsize=(8,5))
plot_tree(
    modelo,
    feature_names=['Temperatura', 'Chuva'],
    class_names=['Não', 'Sim'],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Árvore de Decisão - Devo ir à praia?", fontsize=14)
plt.show()
```
