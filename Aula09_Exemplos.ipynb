## K-Means Clustering

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Gerando dados artificiais com 3 agrupamentos (centros)
X, _ = make_blobs(n_samples=200, centers=3, random_state=42)

# Criando o modelo K-Means com 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42)

# Treinando o modelo e obtendo o rótulo (cluster) de cada ponto
y_kmeans = kmeans.fit_predict(X)

# Plotando os pontos, colorindo conforme o cluster
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis')

# Marcando os centróides dos clusters
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            s=200, c='red', marker='X', label='Centroides')

plt.title('K-Means Clustering')
plt.legend()
plt.show()
```

## Agrupamento Hierárquico (Agglomerative Clustering)

```python
from sklearn.cluster import AgglomerativeClustering
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt

# Cria um dendrograma para visualizar o processo de fusão dos clusters
plt.figure(figsize=(6, 4))
sch.dendrogram(sch.linkage(X, method='ward'))
plt.title('Dendrograma - Hierarchical Clustering')
plt.xlabel('Pontos de dados')
plt.ylabel('Distância')
plt.show()

# Cria o modelo de agrupamento hierárquico com 3 clusters
hier = AgglomerativeClustering(n_clusters=3, linkage='ward')

# Treina o modelo e obtém os rótulos dos clusters
y_hier = hier.fit_predict(X)

# Plotando os pontos de dados agrupados
plt.scatter(X[:, 0], X[:, 1], c=y_hier, cmap='plasma')
plt.title('Hierarchical Clustering')
plt.show()
```

## DBSCAN (Density-Based Spatial Clustering)

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Criando o modelo DBSCAN
# eps: raio de vizinhança; min_samples: mínimo de pontos para formar um cluster
dbscan = DBSCAN(eps=0.7, min_samples=5)

# Treina o modelo e obtém os rótulos dos clusters
y_dbscan = dbscan.fit_predict(X)

# Plotando os pontos, coloridos conforme o cluster
# Pontos com rótulo -1 são considerados "outliers" (ruído)
plt.scatter(X[:, 0], X[:, 1], c=y_dbscan, cmap='rainbow')
plt.title('DBSCAN Clustering')
plt.show()
```
